```python
# Cell 29: Import model training and evaluation libraries

from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

```python
# Cell 30: Create a reusable function to evaluate leak detection models

def evaluate_model(model_name, y_true, y_pred):

    # Calculate performance metrics
    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    # Print model name
    print("=" * 60)
    print(f"{model_name} Evaluation")
    print("=" * 60)

    # Print the main metrics
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-Score:  {f1:.4f}")

    # Print detailed classification report
    print("\nClassification Report:")
    print(
        classification_report(
            y_true,
            y_pred,
            target_names=["No Leak", "Leak"],
            zero_division=0
        )
    )

    # Create confusion matrix
    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    # Display confusion matrix
    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["No Leak", "Leak"]
    )

    disp.plot()

    plt.title(f"{model_name} - Confusion Matrix")
    plt.show()

    # Return results
    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision_Leak": precision,
        "Recall_Leak": recall,
        "F1_Leak": f1
    }
```

```python
# Cell 31: Train baseline Pressure SVM

pressure_svm_baseline = SVC(
    kernel="rbf",
    class_weight="balanced",
    random_state=42
)

# Train using the preprocessed training data
pressure_svm_baseline.fit(
    X_train_pressure_processed,
    y_train
)

# Predict validation data
pressure_validation_pred_baseline = (
    pressure_svm_baseline.predict(
        X_validation_pressure_processed
    )
)

# Evaluate validation performance
pressure_baseline_results = evaluate_model(
    "Pressure SVM - Baseline",
    y_validation,
    pressure_validation_pred_baseline
)
```

```python
# Cell 36: Create a function for validation-based SVM tuning

def tune_svm(
    X_train,
    y_train,
    X_validation,
    y_validation,
    C_values,
    gamma_values,
    model_name
):

    results = []
    best_model = None
    best_f1 = -1

    for C in C_values:

        for gamma in gamma_values:

            # Create SVM
            model = SVC(
                kernel="rbf",
                C=C,
                gamma=gamma,
                class_weight="balanced",
                random_state=42
            )

            # Train on training data only
            model.fit(
                X_train,
                y_train
            )

            # Predict validation data
            y_pred = model.predict(
                X_validation
            )

            # Calculate validation metrics
            accuracy = accuracy_score(
                y_validation,
                y_pred
            )

            precision = precision_score(
                y_validation,
                y_pred,
                pos_label=1,
                zero_division=0
            )

            recall = recall_score(
                y_validation,
                y_pred,
                pos_label=1,
                zero_division=0
            )

            f1 = f1_score(
                y_validation,
                y_pred,
                pos_label=1,
                zero_division=0
            )

            # Save results
            results.append({
                "C": C,
                "gamma": gamma,
                "Accuracy": accuracy,
                "Precision_Leak": precision,
                "Recall_Leak": recall,
                "F1_Leak": f1
            })

            # Keep the best model according to Leak F1
            if f1 > best_f1:

                best_f1 = f1
                best_model = model

                best_parameters = {
                    "C": C,
                    "gamma": gamma
                }

    results_df = pd.DataFrame(results)

    print("=" * 60)
    print(f"Best {model_name} Parameters")
    print("=" * 60)
    print(best_parameters)
    print(f"Best Validation Leak F1: {best_f1:.4f}")

    return best_model, best_parameters, results_df
```

```python
# Cell 37: Define SVM hyperparameter values to test

C_values = [
    0.1,
    1,
    10
]

gamma_values = [
    "scale",
    0.01,
    0.1
]

print("C values:", C_values)
print("Gamma values:", gamma_values)
```

```python
# Cell 38: Tune Pressure SVM

best_pressure_svm, best_pressure_params, pressure_tuning_results = tune_svm(
    X_train_pressure_processed,
    y_train,
    X_validation_pressure_processed,
    y_validation,
    C_values,
    gamma_values,
    "Pressure SVM"
)

display(
    pressure_tuning_results.sort_values(
        by="F1_Leak",
        ascending=False
    )
)
```

```python
# Cell 39: Tune Flow SVM

best_flow_svm, best_flow_params, flow_tuning_results = tune_svm(
    X_train_flow_processed,
    y_train,
    X_validation_flow_processed,
    y_validation,
    C_values,
    gamma_values,
    "Flow SVM"
)

display(
    flow_tuning_results.sort_values(
        by="F1_Leak",
        ascending=False
    )
)
```

```python
# Cell 40: Tune Vibration SVM

best_vibration_svm, best_vibration_params, vibration_tuning_results = tune_svm(
    X_train_vibration_processed,
    y_train,
    X_validation_vibration_processed,
    y_validation,
    C_values,
    gamma_values,
    "Vibration SVM"
)

display(
    vibration_tuning_results.sort_values(
        by="F1_Leak",
        ascending=False
    )
)
```

```python
# Cell 43: Create Hard Majority Voting function

def hard_majority_vote(
    pressure_predictions,
    flow_predictions,
    vibration_predictions
):

    # Combine predictions column-wise
    predictions = np.column_stack([
        pressure_predictions,
        flow_predictions,
        vibration_predictions
    ])

    # Sum the number of "Leak = 1" votes
    leak_votes = predictions.sum(axis=1)

    # Majority vote:
    # 2 or 3 Leak votes  -> Leak (1)
    # 0 or 1 Leak votes  -> No Leak (0)

    final_predictions = (
        leak_votes >= 2
    ).astype(int)

    return final_predictions
```

```python
# Cell 44: Generate Late Fusion predictions on validation data

late_fusion_validation_pred = hard_majority_vote(
    pressure_validation_pred,
    flow_validation_pred,
    vibration_validation_pred
)

# Evaluate the Late Fusion model
late_fusion_validation_results = evaluate_model(
    "Late Fusion - Validation",
    y_validation,
    late_fusion_validation_pred
)
```
